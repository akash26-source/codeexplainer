```python
# ============================================================
# 🩸 BLOOD DONATION ASSISTANT
# LangGraph Project
# Google Colab - Single Cell
# No Gemini API required
# ============================================================

!pip install -q -U langgraph

from typing import TypedDict
from langgraph.graph import StateGraph, START, END


# ------------------------------------------------------------
# 1. State
# ------------------------------------------------------------

class BloodDonationState(TypedDict):
    question: str
    category: str
    answer: str


# ------------------------------------------------------------
# 2. Question Classifier
# ------------------------------------------------------------

def classify_question(state):

    question = state["question"].lower()

    if any(word in question for word in [
        "age", "eligible", "eligibility", "donate",
        "health", "weight", "old"
    ]):
        category = "Eligibility"

    elif any(word in question for word in [
        "blood group", "blood type", "o+", "o-",
        "a+", "a-", "b+", "b-", "ab+", "ab-",
        "compatible", "receive"
    ]):
        category = "Blood Compatibility"

    elif any(word in question for word in [
        "eat", "food", "drink", "before",
        "after", "prepare", "preparation", "water"
    ]):
        category = "Donation Preparation"

    else:
        category = "General Information"

    return {
        "category": category
    }


# ------------------------------------------------------------
# 3. Answer Generator
# ------------------------------------------------------------

def generate_answer(state):

    question = state["question"].lower()
    category = state["category"]

    # Eligibility
    if category == "Eligibility":

        if "age" in question or "old" in question:
            answer = (
                "In many blood donation systems, donors generally "
                "need to be at least 18 years old. Eligibility can "
                "vary by location and blood bank, so the donation "
                "center should confirm the current requirements."
            )

        elif "weight" in question:
            answer = (
                "Minimum weight requirements can vary by blood "
                "donation service. A blood donation center checks "
                "your weight and overall health before donation."
            )

        else:
            answer = (
                "Blood donation eligibility depends on factors such "
                "as age, weight, health condition, medications and "
                "local blood-bank guidelines."
            )


    # Blood Compatibility
    elif category == "Blood Compatibility":

        if "o+" in question:
            answer = (
                "O positive red blood cells can generally be "
                "given to O positive, A positive, B positive and "
                "AB positive recipients."
            )

        elif "o-" in question:
            answer = (
                "O negative red blood cells can generally be "
                "given to people of all ABO and Rh blood groups. "
                "It is commonly called the universal red-cell donor."
            )

        elif "ab+" in question:
            answer = (
                "AB positive is commonly called the universal "
                "red-cell recipient because AB+ individuals can "
                "generally receive red blood cells from all ABO "
                "and Rh groups."
            )

        else:
            answer = (
                "Blood compatibility depends on the ABO and Rh "
                "blood groups. Blood banks perform compatibility "
                "testing before transfusion."
            )


    # Donation Preparation
    elif category == "Donation Preparation":

        if "eat" in question or "food" in question:
            answer = (
                "Before donating blood, eat a normal balanced meal "
                "and drink enough fluids. Follow the instructions "
                "given by your blood donation center."
            )

        elif "water" in question or "drink" in question:
            answer = (
                "Staying well hydrated before and after blood "
                "donation can help you feel better. Follow the "
                "specific instructions from your donation center."
            )

        else:
            answer = (
                "Before donating, have a normal meal, stay hydrated, "
                "get adequate rest and follow the instructions "
                "provided by the blood donation center."
            )


    # General Information
    else:

        if "important" in question or "why" in question:
            answer = (
                "Blood donation is important because donated blood "
                "is used for patients who need transfusions, such "
                "as people undergoing surgery, treatment for certain "
                "illnesses, or recovering from serious injuries. "
                "Regular voluntary donations help maintain blood "
                "supplies."
            )

        elif "how much" in question:
            answer = (
                "The amount collected during a blood donation "
                "depends on local blood-bank procedures and the "
                "type of donation."
            )

        elif "how often" in question:
            answer = (
                "How often you can donate depends on the type of "
                "donation, your health and the rules of the local "
                "blood donation service."
            )

        else:
            answer = (
                "Blood donation is a voluntary process in which "
                "eligible donors give blood that can be tested, "
                "processed and used for patients who need "
                "transfusions."
            )

    return {
        "answer": answer
    }


# ------------------------------------------------------------
# 4. Create LangGraph
# ------------------------------------------------------------

workflow = StateGraph(BloodDonationState)

workflow.add_node(
    "classifier",
    classify_question
)

workflow.add_node(
    "answer_generator",
    generate_answer
)

workflow.add_edge(
    START,
    "classifier"
)

workflow.add_edge(
    "classifier",
    "answer_generator"
)

workflow.add_edge(
    "answer_generator",
    END
)

app = workflow.compile()


# ------------------------------------------------------------
# 5. Run Assistant
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("        🩸 BLOOD DONATION ASSISTANT")
print("=" * 60)

question = input("\nEnter your question: ")

result = app.invoke({
    "question": question,
    "category": "",
    "answer": ""
})


# ------------------------------------------------------------
# 6. Display Result
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("CATEGORY:", result["category"])
print("-" * 60)
print("ASSISTANT:")
print(result["answer"])
print("-" * 60)

```
