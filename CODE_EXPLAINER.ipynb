```python
import os
from getpass import getpass
from google import genai

# Enter Gemini API key
os.environ["GEMINI_API_KEY"] = getpass("Enter your Gemini API key: ")

# Create Gemini client
client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])

# Get programming language
language = input("Enter programming language: ")

print("Paste your code. Type END when finished:")

code_lines = []

while True:
    line = input()

    if line.strip() == "END":
        break

    code_lines.append(line)

code = "\n".join(code_lines)

# Prompt
prompt = f"""
You are a programming teacher.

Explain this {language} code to a beginner.

Include:
1. What the code does
2. Line-by-line explanation
3. Expected output
4. Possible errors
5. Improvements

Code:

{code}
"""

# Ask Gemini
response = client.models.generate_content(
    model="gemini-3.6-flash", # Updated model name to gemini-3.6-flash
    contents=prompt
)

print("\n========== CODE EXPLANATION ==========\n")
print(response.text)
```
